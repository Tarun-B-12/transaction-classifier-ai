```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../data/raw/transactions.csv")

print("Rows:", len(df))
print("Columns:", list(df.columns))
print()
df.head(10)
```

```python
print("Category counts:")
print(df["true_category"].value_counts())
print()
print("Number of categories:", df["true_category"].nunique())
```

```python
print("Sample descriptions (the text our classifier must interpret):")
for desc in df["description"].sample(15, random_state=1):
    print("  ", repr(desc))
```

```python
print("Missing values per column:")
print(df.isnull().sum())
print()
print("Duplicate rows:", df.duplicated().sum())
print()
print("Amount range: $", df["amount"].min(), "to $", df["amount"].max())
```

## Day 1 — Data Exploration Notes

- 2,000 transactions across 9 spend categories, roughly balanced (184–270 each).
- Descriptions are messy in realistic ways: inconsistent casing, doubled/trailing
  spaces, payment-processor prefixes (TST*, SQ*, POS DEBIT), and store numbers.
- No missing values or duplicates — data quality is clean, so the work is text
  classification, not data repair.
- Built-in ambiguity to watch: "UBER* EATS" is labeled Transport but is really
  food delivery — a known hard case for the evaluation.
